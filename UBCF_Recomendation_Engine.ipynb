# 2.Buildinf UBCH using euclideon  distance

```python
from sklearn.metrics import pairwise_distances
```

```python
euclidion_piv_table=pd.pivot_table(data=movies_data,values="rating",index="userId",columns="movie").fillna(0)
euclidion_piv_table.columns=movies_data["movie"].unique()
euclidion_piv_table.index=movies_data["userId"].unique()
euclidion_piv_table
```

```python
eulidion_correlation=pd.DataFrame(pairwise_distances(X=euclidion_piv_table,metric='euclidean'))
eulidion_correlation.columns=movies_data["userId"].unique()
eulidion_correlation.index=movies_data["userId"].unique()
eulidion_correlation.round(2)
```

# 3.Buildinf UBCH using cosine similarity

```python
cosine_correlation=pd.DataFrame(1-pairwise_distances(X=euclidion_piv_table,metric='cosine'))
cosine_correlation.columns=movies_data["userId"].unique()
cosine_correlation.index=movies_data["userId"].unique()
cosine_correlation.round(2)
```
